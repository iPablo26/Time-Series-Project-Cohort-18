```python


# 1. Sample Setup: Create a DataFrame with daily temperature observations
# Assume 'df' has a DatetimeIndex and a 'temperature' column
dates = pd.date_range(start='2025-01-01', end='2025-12-31', freq='D')

np.random.seed(42)
temperatures = np.random.uniform(low=10.0, high=35.0, size=len(dates))

df = pd.DataFrame({'temperature': temperatures}, index=dates)

# 2. Resample into Weekly Averages ('W')
# 'W' groups data by week (defaulting to Sunday ending)
weekly_avg = df.resample('W').mean()

# 3. Resample into Monthly Averages ('M')
# 'M' groups data by month-end frequency
monthly_avg = df.resample('M').mean()

print("--- Weekly Temperature Averages ---")
print(weekly_avg.head())

print("\n--- Monthly Temperature Averages ---")
print(monthly_avg)
```
