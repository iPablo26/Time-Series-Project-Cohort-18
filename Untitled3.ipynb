```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
pd.read_csv=("noaa-5000.csv")
df = pd.DataFrame({
    "date": ["2025-01-01", "2025-01-02", "2025-01-03"],
    "temperature": [25.4, 26.1, 24.8]
})

# Convert date strings to datetime objects
df["date"] = pd.to_datetime(df["date"])

print(df)
print(df.dtypes)
```

```python
df["year"]=df['date'].dt.year
df["month"]=df["date"].dt.month
```

```python
dates = pd.date_range(
    start="2025-01-01",
    end="2025-12-31",
    freq="D"
)

print(dates)
print("Number of days:", len(dates))

# Set the date column as the index
df = df.set_index("date")
```

```python
np.random.seed(42)

df = pd.DataFrame(
    {"temperature": 25 + 5 * np.sin(
        2 * np.pi * np.arange(len(dates)) / 365
    ) + np.random.normal(0, 1.5, len(dates))},
    index=dates
)

df.index.name = "date"
```

```python
# All observations from July 2025
july_data = df.loc["2025-07"]

# Observations between two dates
first_week = df.loc["2025-01-01":"2025-01-07"]
```

```python
# Weekly average temperature
weekly_avg = df["temperature"].resample("W").mean()

# Monthly average temperature
monthly_avg = df["temperature"].resample("ME").mean()

print("Weekly averages:")
print(weekly_avg.head())

print("\nMonthly averages:")
print(monthly_avg)
```

```python
df["rolling_30_day"] = (
    df["temperature"]
    .rolling(window=30, min_periods=30)
    .mean()
)

print(df.head(35))
```

```python
# Extract calendar attributes
df["month"] = df.index.month
df["day_of_week"] = df.index.dayofweek
df["day_name"] = df.index.day_name()

# Average temperature by month
monthly_pattern = df.groupby("month")["temperature"].mean()
```

```python
monthly_names = (
    df.groupby(df.index.month)["temperature"]
    .mean()
)

monthly_names.index = [
    pd.Timestamp(2025, month, 1).strftime("%B")
    for month in monthly_names.index
]
```

```python
# Prepare the data
daily = df["temperature"]
rolling = df["rolling_30_day"]
monthly = df["temperature"].resample("ME").mean()

# Create the plot
plt.figure(figsize=(14, 6))

plt.plot(
    daily.index,
    daily,
    label="Daily Temperature",
    alpha=0.35,
    linewidth=1
)

plt.plot(
    rolling.index,
    rolling,
    label="30-Day Rolling Average",
    linewidth=2
)

plt.plot(
    monthly.index,
    monthly,
    label="Monthly Average",
    marker="o",
    linewidth=2
)

plt.title("Daily Temperature and Time-Series Trends (2025)")
plt.xlabel("Date")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
