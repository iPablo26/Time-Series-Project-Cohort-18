```python

import pandas as pd
import numpy as np

# Step 1: Create daily dates for January 2025
dates = pd.date_range(
    start='2025-01-01',
    end='2025-01-31',
    freq='D'
)

# Step 2: Create sample temperature data
df = pd.DataFrame({
    'Date': dates,
    'Temperature': np.random.randint(25, 36, size=len(dates))
})

# Step 3: Set Date as the index
df = df.set_index('Date')

# Step 4: Display daily temperatures
print("Daily Temperatures:")
print(df)

# Step 5: Calculate weekly average temperatures
weekly_avg = df['Temperature'].resample('W').mean()

print("\nWeekly Average Temperatures:")
print(weekly_avg)

# Step 6: Calculate monthly average temperatures
monthly_avg = df['Temperature'].resample('ME').mean()

print("\nMonthly Average Temperatures:")
print(monthly_avg)
```

```python

import pandas as pd
import numpy as np

# Step 1: Create daily dates for 2025
dates = pd.date_range(
    start='2025-01-01',
    periods=90,
    freq='D'
)

# Step 2: Create sample temperature data
np.random.seed(10)

df = pd.DataFrame({
    'Date': dates,
    'Temperature': np.random.randint(25, 36, size=len(dates))
})

# Step 3: Set Date as the index
df = df.set_index('Date')

# Step 4: Calculate the 30-day rolling average
df['Rolling_Average'] = (
    df['Temperature'].rolling(window=30).mean()
)

# Step 5: Display the results
print("Daily Temperatures and 30-Day Rolling Average:")
print(df)

# Step 6: Display the last 10 rows
print("\nLast 10 Rows:")
print(df.tail(10))
```

```python

import pandas as pd
import numpy as np

# Step 1: Create daily dates for 2025
dates = pd.date_range(
    start='2025-01-01',
    periods=90,
    freq='D'
)

# Step 2: Create sample temperature data
np.random.seed(10)

df = pd.DataFrame({
    'Date': dates,
    'Temperature': np.random.randint(25, 36, size=len(dates))
})

# Step 3: Set Date as the index
df = df.set_index('Date')

# Step 4: Extract calendar attributes
df['Year'] = df.index.year
df['Month'] = df.index.month
df['Day'] = df.index.day
df['Day_of_Week'] = df.index.day_name()

# Step 5: Display the dataset
print("Dataset with Calendar Attributes:")
print(df)

# Step 6: Calculate average temperature by month
monthly_average = df.groupby('Month')['Temperature'].mean()

print("\nAverage Temperature by Month:")
print(monthly_average)

# Step 7: Calculate average temperature by day of the week
weekday_average = df.groupby('Day_of_Week')['Temperature'].mean()

print("\nAverage Temperature by Day of the Week:")
print(weekday_average)
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Create daily dates
dates = pd.date_range(
    start='2025-01-01',
    periods=365,
    freq='D'
)

# Step 2: Create sample temperature data
np.random.seed(10)

df = pd.DataFrame({
    'Date': dates,
    'Temperature': (
        28
        + 5 * np.sin(2 * np.pi * dates.dayofyear / 365)
        + np.random.normal(0, 2, len(dates))
    )
})

# Step 3: Set Date as the index
df = df.set_index('Date')

# Step 4: Calculate the 30-day rolling average
df['Rolling_Average'] = (
    df['Temperature'].rolling(window=30).mean()
)

# Step 5: Calculate monthly average temperatures
monthly_avg = df['Temperature'].resample('ME').mean()

# Step 6: Plot the data
plt.figure(figsize=(14, 7))

plt.plot(
    df.index,
    df['Temperature'],
    label='Daily Temperature',
    alpha=0.4
)

plt.plot(
    df.index,
    df['Rolling_Average'],
    label='30-Day Rolling Average',
    linewidth=2
)

plt.plot(
    monthly_avg.index,
    monthly_avg.values,
    label='Monthly Average',
    marker='o',
    linewidth=2
)

# Step 7: Add labels and title
plt.title('Temperature Time Series Analysis - 2025')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Step 8: Display the graph
plt.show()
```
