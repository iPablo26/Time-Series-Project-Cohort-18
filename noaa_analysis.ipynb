```python
df = pd.read_csv("noaa - 5000.csv")

df
```

```python
df.head
```

```python
df["date"] = pd.to_datetime(df["date"])

df = df.set_index("date")

df
```

```python

df.columns
```

```python

df["temp"].resample("ME").mean()
```

```python

weekly_avg = df["temp"].resample("W").mean()

weekly_avg
```

```python

df["rolling_30_day"] = df["temp"].rolling(window=30).mean()

df[["temp", "rolling_30_day"]].head(35)
```

```python
df[["temp", "rolling_30_day"]].head(35)
```

```python

df["month"] = df.index.month

df[["temp", "month"]].head()
```

```python

df["day_of_week"] = df.index.dayofweek

df[["temp", "month", "day_of_week"]].head()
```

```python

monthly_seasonal_avg = df.groupby("month")["temp"].mean()

monthly_seasonal_avg
```

```python
import matplotlib.pyplot as plt
```

```python

plt.figure(figsize=(14, 6))

plt.plot(df.index, df["temp"], label="Daily Temperature", alpha=0.4)
plt.plot(df.index, df["rolling_30_day"], label="30-Day Rolling Average")

monthly_avg = df["temp"].resample("ME").mean()
plt.plot(monthly_avg.index, monthly_avg, marker="o", label="Monthly Average")

plt.title("NOAA Temperature Time Series Analysis")
plt.xlabel("Date")
plt.ylabel("Temperature")
plt.legend()
plt.grid(True)
plt.show()
```
